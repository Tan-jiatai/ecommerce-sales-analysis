```python
import pandas as pd
import numpy as np
```

<u>**STEP 1 : LOAD DATA**</u>

```python
df = pd.read_csv('ecommerce_sales_34500.csv')
#Explanation : Reads the CSV file and stores it as a DataFrame called 'df'
```

<ins>**STEP 2 ：EXPLORE DATA**</ins>

```python
print("=== ORIGINAL DATA ===")
print(f"Shape:{df.shape}")
#Explanation : Shows (rows, columns) - e.g., (34500, 12) means 34,500 rows and 12 columns
```

```python
print(f"Columns: {df.columns.tolist()}")
#Explantion : Lists all column names in the dataset
```

```python
print("\nFirst 5 rows:")
print(df.head())
#Explanation : Shows first 5 rows to see what the data looks like
```

```python
print("\nMissing values:")
print(df.isnull().sum())
#Explantion ：Counts how many missing values in each column
```

```python
print(f"\nDuplicates: {df.duplicated().sum()}")
#Explanation : Counts how many exact duplicate rows exist
```

<u>**STEP 3: CLEAN DATA**</u>

```python
df = df.drop_duplicates()
#Explanation : Removes all duplicate rows (keeps only first occurrence)

df = df.dropna()
#Explanation :Removes any row that has at least one missing value
```

<u>**STEP 4: CONVERT DATE COLUMN**</u>

```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
# Explanation :
# Converts text dates (e.g., "2024-03-15") to actual date format
# errors='coerce' means if conversion fails, make it NaN instead of crashing

df['year'] = df['order_date'].dt.year
# Explanation : # Extracts year from date, Example : 2024-03-15 → 2024

df['month'] = df['order_date'].dt.month
# Explanation : # Extracts month from date, Example : 2024-03-15 → 3

df['day'] = df['order_date'].dt.day
# Explanation : # Extracts day from date, Example : 2024-03-15 → 15

df['weekday'] = df['order_date'].dt.day_name()
# Explanation : # Extracts day name from date, Example : 2024-03-15 → Friday

df['quarter'] = df['order_date'].dt.quarter
# Explanation : # Extracts quater from date, Example : 2024-03-15 →  1 (Jan-Mar = Q1, Apr-Jun = Q2, etc.)
```

<u>**STEP 5: CONVERT NUMERIC COLUMNS**</u>

```python
df['price'] = pd.to_numeric(df['price'], errors='coerce')
# Explantion :
# Converts text like "$100" or "100.50" to actual number 100.50
# errors='coerce' means if conversion fails, make it NaN

df['total_amount'] = pd.to_numeric(df['total_amount'], errors='coerce')
# Explanation : Same for total_amount column

df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')
# Explanation : Same for quantity col6mn
```

<u>**STEP 6: REMOVE INVALID ROWS**</u>

```python
df = df[df['quantity'] > 0]
# Explantion :
# Keeps only rows where quantity is greater than 0
# Removes rows with quantity = 0 or negative (returns/refunds)

df = df[df['total_amount'] > 0]
# Explantion :
# Keeps only rows where total_amount is greater than 0
# Removes rows with zero or negative sales
```

<u>**STEP 7: CREATE NEW COLUMN**</u>

```python
df['unit_price'] = df['total_amount'] / df['quantity']
# Explantion :
# Calculates price per unit by dividing total amount by quantity
# Example: total_amount = 100, quantity = 5 → unit_price = 20
```

<u>**STEP 8: RESET INDEX**</u>

```python
df = df.reset_index(drop=True)
# Explanation : 
# Renumbers rows from 0, 1, 2, 3... after deleting rows
# drop=True means don't keep old index as a column
```

<u>**STEP 9: SAVE CLEANED DATA**</u>

```python
df.to_csv('ecommerce_cleaned.csv', index=False)
# Explanation:
# Saves cleaned DataFrame to CSV file
# index=False means don't save row numbers in the file
```

<u>**STEP 10: VERIFY CLEANED DATA**</u>

```python
print(f"Final shape: {df.shape}")
# Explanation: Shows how many rows and columns after cleaning
```

```python
print("\nFirst 5 rows:")
print(df.head())
# Explanation: Shows first 5 rows of cleaned data
```

```python
print("\nSummary statistics:")
print(df.describe())
# Explanation : Shows statistics for numeric columns:
# - count: how many values
# - mean: average
# - std: standard deviation
# - min: minimum value
# - 25%, 50%, 75%: percentiles
# - max: maximum value
```

<u>**STEP 11: QUICK BUSINESS SUMMARY**</u>

```python
print(f"Total Sales: ${df['total_amount'].sum():,.2f}")
# Explanation :
# Sums all total_amount values → total revenue
# :,.2f formats as currency with commas and 2 decimals

print(f"Total Orders: {df['order_id'].nunique()}")
# Explanation :Counts unique order IDs → total number of orders

print(f"Total Customers: {df['customer_id'].nunique()}")
# Explanation :Counts unique customer IDs → total number of customers

print(f"Categories: {df['category'].nunique()}")
# Explanation :Counts unique categories → how many product categories

print(f"Regions: {df['region'].nunique()}")
# Explanation :Counts unique regions → how many regions/countries
```
